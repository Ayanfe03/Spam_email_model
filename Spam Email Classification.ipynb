```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
raw_data = pd.read_csv('mail_data.csv')
```

```python
raw_data.head()
```

```python
mail_data = raw_data.where((pd.notnull(raw_data)),'')
```

```python
mail_data.head()
```

```python
mail_data.shape
```

```python
mail_data.loc[mail_data['Category'] == 'spam', 'Category'] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category'] = 1
```

```python
mail_data['Category'].value_counts()
```

```python
ham_data = mail_data[mail_data.Category == 1]
spam_data = mail_data[mail_data.Category == 0]
```

```python
ham_data.shape
```

```python
spam_data.shape
```

```python
ham_sample = ham_data.sample(n=747, random_state=1)
ham_sample.shape
```

```python
new_mail_data = pd.concat([ham_sample, spam_data], axis=0)
new_mail_data['Category'].value_counts()
```

```python
X = new_mail_data['Message']
Y = new_mail_data['Category']
new_mail_data.shape
```
