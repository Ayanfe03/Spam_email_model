#Spam Mail Prediction using Logistic Regression Model


```python
# libraries imported in earlier projects will already be active thoughout the entire notebook
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
# load the csv file into a pandas dataframe
raw_data = pd.read_csv('mail_data.csv')
```

```python
raw_data.head()
```

```python
# replacing all the null values with null string
mail_data = raw_data.where((pd.notnull(raw_data)),'')
```

```python
mail_data.head()
```

```python
mail_data
```

```python
# label encoding
mail_data.loc[mail_data['Category'] == 'spam', 'Category'] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category'] = 1
```

```python
mail_data['Category'].value_counts()
```

```python
ham_data = mail_data[mail_data.Category == 1]
spam_data = mail_data[mail_data.Category == 0]
```

```python
ham_data.shape
```

```python
spam_data.shape
```

```python
# using undersampling to balance the dataset
ham_sample = ham_data.sample(n=747, random_state=1)
ham_sample.shape
```

```python
new_mail_data = pd.concat([ham_sample, spam_data], axis=0)
new_mail_data['Category'].value_counts()
```

```python
X = new_mail_data['Message']
Y = new_mail_data['Category']
new_mail_data.shape
```

```python
print(X)
print(Y)
```

```python
# splitting dataset into train and test sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
# feature_extraction for train and test features
feature_extract = CountVectorizer(min_df = 1, stop_words='english', lowercase=True)

feature_extract.fit(X_train)
X_train_features = feature_extract.transform(X_train)
X_test_features = feature_extract.transform(X_test)

Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

```python
print(X_test_features)
```

```python
log_model = LogisticRegression()
```

```python
# fitting training data to instance of Logistic Regression model
log_model.fit(X_train_features, Y_train)
```

```python
# prediction on training data
train_pred = log_model.predict(X_train_features)
accuracy_train = accuracy_score(train_pred, Y_train)
```

```python
print('Accuracy on training data is:', accuracy_train)
print('Precision on training data:', precision_score(train_pred, Y_train))
print('Recall on training data:', recall_score(train_pred, Y_train))
```

```python
# prediction on test data
test_pred = log_model.predict(X_test_features)
accuracy_test = accuracy_score(test_pred, Y_test)
```

```python
print('Accuracy on test data is:', accuracy_test)
print('Precision on test data:', precision_score(test_pred, Y_test))
print('Recall on test data:', recall_score(test_pred, Y_test))
```

```python
# building a predictive system
input_mail = ["I'm learning a lot about Data Science and ML Basics with GrowthPro Africa!"]
input_mail_features = feature_extract.transform(input_mail)

prediction = log_model.predict(input_mail_features)
print(prediction)

if prediction[0] == 1:
    print("Ham mail")
else:
    print("Spam mail")
```
